```python
import pandas as pd
df = pd.read_csv("../data/weather.csv")
df.info()
df.describe()
```

```python
import matplotlib.pyplot as plt
df["time"] = pd.to_datetime(df["time"])
df.set_index("time")["temperature_2m_max"].plot(figsize=(10,4), title="Daily max temperature")
plt.show() 
```

```python
df["month"] = df["time"].dt.month
df.groupby("month")["precipitation_sum"].mean().plot(kind="bar", figsize=(10,4), title="Average daily rainfall by month (mm)")
plt.show()
```

```python
df.set_index("time")["et0_fao_evapotranspiration"].plot(figsize=(10,4), title="Daily ET0 (mm/day)")
plt.show()
```
