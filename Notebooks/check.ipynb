```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/dataset.csv")
df["time"] = pd.to_datetime(df["time"])

print(df.groupby("soil")["irrigate"].sum())   # हर मिट्टी में सिंचाई के दिन

s = df[(df["soil"] == "loamy") & (df["time"].dt.year == 2023)]
s = s.set_index("time")
fig, ax = plt.subplots(figsize=(10, 4))
s["soil_moisture"].plot(ax=ax, label="Soil moisture (mm)")
ax.scatter(s.index[s["irrigate"] == 1], s["soil_moisture"][s["irrigate"] == 1],
           color="red", label="Irrigation day")
ax.set_title("Loamy soil, wheat season 2022-23")
ax.legend()
plt.show() 
```
